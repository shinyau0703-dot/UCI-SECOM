```python
import requests,pandas as pd
from io import BytesIO

def uci_data_request(train_url,labels_url):
  r_train=requests.get(train_url)
  train=pd.read_csv(BytesIO(r_train.content),sep=" ",header=None)
  r_labels=requests.get(labels_url)
  labels=pd.read_csv(BytesIO(r_labels.content),sep=" ",header=None)
  df=pd.merge(train,labels,left_index=True,right_index=True,how='inner')
  df.rename(columns={'0_x':'0','1_x':'1','0_y':'Pass/Fail','1_y':'Time'},inplace=True)
  return train,labels,df

train_url="https://archive.ics.uci.edu/ml/machine-learning-databases/secom/secom.data"
labels_url="https://archive.ics.uci.edu/ml/machine-learning-databases/secom/secom_labels.data"

train,labels,df=uci_data_request(train_url,labels_url)

out_dir=r"D:\Sandy\UCI-SECOM\data"
train.to_csv(out_dir+r"\secom_train.csv",index=False)
labels.to_csv(out_dir+r"\secom_labels.csv",index=False)
df.to_csv(out_dir+r"\secom_merged.csv",index=False)
```
